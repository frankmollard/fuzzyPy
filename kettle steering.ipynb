```python
import numpy as np 
from fuzzyfier import fuzzyfication
```

If there should be just one result e.g. a ketlle outlet has to be steered, a defuzzsification is required.

```python
pressureBar = {"very low": 0, "low": 0.5, "medium": 1, "high": 2, "very high": 3}
temperature = {"very low": 0, "low": 15, "medium": 50, "high": 100}
```

```python
pB=fuzzyfication(pressureBar, 1.4)
pT=fuzzyfication(temperature, 65)
```

Checking the activations of both values 1.4 and 65

```python
pB.fuzzy()
```

```python
pT.fuzzy()
```

Now we calculate the conditions:

* If pressure high or temperature high: open outlet of kettle. Condition 1
* If pressure is very low: open outlet. Cause kettle is empty. Condition 2

```python
Condition1 = np.max([pT.fuzzy()["high"], pB.fuzzy()["high"]])
Condition2 = pB.fuzzy()["very low"]
```

For defuzzyfication, we create a dictionare with all relevant conditions

```python
decision = {"Condition1": Condition1, "Condition2": Condition2}
```

The retulting values are often dependent from the activations. So, we calculate the percentage of the outlet as a magnitude of openness. We can use the default method minimum value of maximal activations.

```python
outlet = pT.defuzzification(decision, [100*Condition1, 100*Condition2])
print("Openness of outlet", np.round(outlet))
```

Or we use weighted mean.

```python
outlet = pT.defuzzification(decision, [100*Condition1, 100*Condition2], method="weightedMean")
print("Openness of outlet", np.round(outlet))
```
